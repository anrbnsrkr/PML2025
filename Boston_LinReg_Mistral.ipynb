```python
import numpy as np
import pymc as pm
import pandas as pd
import arviz as az
# from sklearn.datasets import load_diabetes  # Example regression dataset
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt


boston = pd.read_csv("BostonHousing.csv")
features_to_rename = {
        "crim"   : "per capita crime rate by town",
         "zn"    : "proportion of residential land zoned for lots over 25,000 sq.ft.",
         "indus" :  "proportion of non-retail business acres per town",
         "chas"  :  "Charles River dummy variable (= 1 if tract bounds river; 0 otherwise)",
         "nox"   :  "nitric oxides concentration (parts per 10 million)",
         "RM"      : "average number of rooms per dwelling",
         "rm"    :  "proportion of owner-occupied units built prior to 1940",
         "age"      :"weighted distances to five Boston employment centres",
         "dis"      :"index of accessibility to radial highways",
         "tax"     : "full-value property-tax rate per $10,000",
         "ptratio" :  "pupil-teacher ratio by town",
         "b"       : "1000(Bk - 0.63)^2 where Bk is the proportion of blacks by town",
         "lstat"    :"% lower status of the population",
         "medv"     :"Median value of owner-occupied homes in $1000's",
    }
boston = boston.drop("rad",axis = 1)
boston = boston.drop("chas",axis = 1)
our_y = boston['medv']
boston = boston.drop("medv",axis=1)

# data = boston.rename(columns=features_to_rename)
# missing = our_y.isnull().values | data.isnull().any(axis=1).values
# data = data[~missing]
# our_y = our_y[~missing]

# data, target, frame, target_names, feature_names, categorical_features = (
#         data,
#         our_y,
#         data.assign(target=our_y),
#         np.array(["no heart disease", "heart disease"]),
#         data.columns.to_numpy(),
#         [],
#     )

# Load regression dataset
# data_bunch = 
# X = data_bunch.data
# y = data_bunch.target

X = boston
y = np.array(our_y)

# Define continuous feature indices (assumed all are continuous here)
# continuous_idx = list(range(X.shape[1]))

# Imputation on continuous features
X_imputed = X.copy()
imputer = SimpleImputer(strategy="mean")
X_imputed = imputer.fit_transform(X)

# Scale continuous features
X_scaled = X_imputed.copy()
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_imputed)

# Train-test split
X_train_full, X_test, y_train_full, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

# Load mixture priors
mixture_params_all = np.load("Boston_Minstral.npy")
assert mixture_params_all.ndim == 3 and mixture_params_all.shape[2] == 2
n_tasks, n_features, _ = mixture_params_all.shape

# Mixture prior model
def build_model_mixture(X, y):
    with pm.Model() as model:
        w_components = []
        for j in range(X.shape[1]):
            comp_dists = [
                pm.Normal.dist(mu=mixture_params_all[t, j, 0],
                               sigma=mixture_params_all[t, j, 1])
                for t in range(n_tasks)
            ]
            w_j = pm.Mixture(f"w_{j}", w=np.ones(n_tasks)/n_tasks, comp_dists=comp_dists)
            w_components.append(w_j)
        w = pm.Deterministic("w", pm.math.stack(w_components))
        b = pm.Normal("b", mu=0, sigma=1)
        mu = pm.math.dot(X, w) + b
        pm.Normal("y_obs", mu=mu, sigma=1, observed=y)
    return model

# Standard normal prior model
def build_model_normal(X, y):
    with pm.Model() as model:
        w = pm.Normal("w", mu=0, sigma=1, shape=X.shape[1])
        b = pm.Normal("b", mu=0, sigma=1)
        mu = pm.math.dot(X, w) + b
        pm.Normal("y_obs", mu=mu, sigma=1, observed=y)
    return model

# Inference
def run_inference(model):
    with model:
        trace = pm.sample(1000, tune=500, chains=2, target_accept=0.9,
                          return_inferencedata=True, progressbar=False)
    return trace

# Predict from posterior
def predict_posterior(trace, X):
    w_samples = trace.posterior['w'].stack(draws=("chain", "draw")).values
    b_samples = trace.posterior['b'].stack(draws=("chain", "draw")).values
    preds = np.einsum('nd,dz->nz', X, w_samples) + b_samples
    return preds.mean(axis=1)

# Evaluation loop
N_DATA_POINTS_SEEN = [2, 5, 10, 20, 30, 40, 50]
mse_results_std = []
mse_results_mix = []

for n in N_DATA_POINTS_SEEN:
    print(f"\nTraining with {n} samples...")

    idx = np.random.choice(len(X_train_full), n, replace=False)
    X_sub = X_train_full[idx]
    y_sub = y_train_full[idx]

    # Standard normal prior
    model_std = build_model_normal(X_sub, y_sub)
    trace_std = run_inference(model_std)
    y_pred_std = predict_posterior(trace_std, X_test)
    mse_std = mean_squared_error(y_test, y_pred_std)
    mse_results_std.append(mse_std)

    # Mixture prior
    model_mix = build_model_mixture(X_sub, y_sub)
    trace_mix = run_inference(model_mix)
    y_pred_mix = predict_posterior(trace_mix, X_test)
    mse_mix = mean_squared_error(y_test, y_pred_mix)
    mse_results_mix.append(mse_mix)

    print(f"MSE (Standard): {mse_std:.2f} | MSE (Mixture): {mse_mix:.2f}")

# Plot MSE results
plt.figure(figsize=(8, 5))
plt.plot(N_DATA_POINTS_SEEN, mse_results_std, marker='o', label='Standard Normal Prior')
plt.plot(N_DATA_POINTS_SEEN, mse_results_mix, marker='x', label='Mixture Prior')
plt.xlabel("Number of Training Data Points")
plt.ylabel("Test MSE")
plt.title("Test MSE vs. Training Data Size")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
