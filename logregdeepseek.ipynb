```python
import numpy as np
import pymc as pm
import arviz as az
# from sklearn.datasets import load_heart_disease  # Replace this with your actual loader
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

from llm_elicited_priors.datasets import (
    load_breast_cancer,load_heart_disease,load_sk_diabetes,load_hypothyroid,load_california_housing
)

# Load data
data_bunch = load_heart_disease()  # Replace if needed
X = data_bunch.data
y = data_bunch.target

# Define which features are continuous
categorical_idx = [1, 4, 5, 7, 9]  # Update these indices as needed
continuous_idx = [i for i in range(X.shape[1]) if i not in categorical_idx]

# Impute missing values in continuous features
X_imputed = X.copy()
imputer = SimpleImputer(strategy="mean")
X_imputed[:, continuous_idx] = imputer.fit_transform(X[:, continuous_idx])

# Scale continuous features
X_scaled = X_imputed.copy()
scaler = StandardScaler()
X_scaled[:, continuous_idx] = scaler.fit_transform(X_imputed[:, continuous_idx])

# Train-test split
X_train_full, X_test, y_train_full, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

# Load mixture priors
mixture_params_all = np.load("our_priors/Heart_gpt4omini.npy")
assert mixture_params_all.ndim == 3 and mixture_params_all.shape[2] == 2
n_tasks, n_features, _ = mixture_params_all.shape

# Define mixture prior model
def build_model_mixture(X, y):
    with pm.Model() as model:
        w_components = []
        for j in range(X.shape[1]):
            comp_dists = [
                pm.Normal.dist(mu=mixture_params_all[t, j, 0],
                               sigma=mixture_params_all[t, j, 1])
                for t in range(n_tasks)
            ]
            w_j = pm.Mixture(f"w_{j}", w=np.ones(n_tasks)/n_tasks, comp_dists=comp_dists)
            w_components.append(w_j)
        w = pm.Deterministic("w", pm.math.stack(w_components))
        b = pm.Normal("b", mu=0, sigma=1)
        logits = pm.math.dot(X, w) + b
        pm.Bernoulli("y_obs", logit_p=logits, observed=y)
    return model

# Define standard normal prior model
def build_model_normal(X, y):
    with pm.Model() as model:
        w = pm.Normal("w", mu=0, sigma=1, shape=X.shape[1])
        b = pm.Normal("b", mu=0, sigma=1)
        logits = pm.math.dot(X, w) + b
        pm.Bernoulli("y_obs", logit_p=logits, observed=y)
    return model

# Inference
def run_inference(model):
    with model:
        trace = pm.sample(5000, tune=1000, chains=5, target_accept=0.95,
                          return_inferencedata=True, progressbar=False)
    return trace

# Predict from posterior
def predict_posterior(trace, X):
    w_samples = trace.posterior['w'].stack(draws=("chain", "draw")).values
    b_samples = trace.posterior['b'].stack(draws=("chain", "draw")).values
    logits = np.einsum('nd,dz->nz', X, w_samples) + b_samples
    probs = 1 / (1 + np.exp(-logits))
    return probs.mean(axis=1)

# Evaluation loop
N_DATA_POINTS_SEEN = [2, 5, 10, 20, 30, 40, 50]
auc_results_std = []
auc_results_mix = []

for n in N_DATA_POINTS_SEEN:
    print(f"\nTraining with {n} samples...")

    idx = np.random.choice(len(X_train_full), n, replace=False)
    X_sub = X_train_full[idx]
    y_sub = y_train_full[idx]

    # Standard normal prior
    model_std = build_model_normal(X_sub, y_sub)
    trace_std = run_inference(model_std)
    y_pred_std = predict_posterior(trace_std, X_test)
    auc_std = roc_auc_score(y_test, y_pred_std)
    auc_results_std.append(auc_std)

    # Mixture prior
    model_mix = build_model_mixture(X_sub, y_sub)
    trace_mix = run_inference(model_mix)
    y_pred_mix = predict_posterior(trace_mix, X_test)
    auc_mix = roc_auc_score(y_test, y_pred_mix)
    auc_results_mix.append(auc_mix)

    print(f"AUC (Standard): {auc_std:.4f} | AUC (Mixture): {auc_mix:.4f}")

# Plot AUC results
plt.figure(figsize=(8, 5))
plt.plot(N_DATA_POINTS_SEEN, auc_results_std, marker='o', label='Standard Normal Prior')
plt.plot(N_DATA_POINTS_SEEN, auc_results_mix, marker='x', label='Mixture Prior')
plt.xlabel("Number of Training Data Points")
plt.ylabel("Test AUC")
plt.title("Test AUC vs. Training Data Size")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
